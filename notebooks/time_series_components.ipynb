# Time Series Components Analysis
Course-based methods only: time plot, ACF, ADF, KPSS, and periodogram.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.stattools import adfuller, kpss
from statsmodels.graphics.tsaplots import plot_acf
from scipy.signal import periodogram
```

## Automobiles NL

```python
ts = pd.read_excel("../data/Autos_NL.xlsx")
print(ts.shape)
print(ts.dtypes)
ts.head()
```

```python
ts["AÑO"] = pd.to_datetime(ts["AÑO"].astype(str), format="%Y")
ts = ts.set_index("AÑO").sort_index()
```

```python
plt.figure(figsize=(14,4))
plt.plot(ts["Nuevo León"])
plt.xticks(rotation=45)
plt.xlabel("Time")
plt.ylabel("Nuevo León")
plt.title("Time series - Automobiles NL")
plt.show()
```

```python
plot_acf(ts["Nuevo León"], lags=min(50, len(ts)//2-1))
plt.xlabel("Lags")
plt.ylabel("Autocorrelation")
plt.title("ACF - Automobiles NL")
plt.show()
```

```python
adf_test = adfuller(ts["Nuevo León"], autolag="AIC")
print("ADF p-value:", adf_test[1])
```

```python
stat, p_value, _, _ = kpss(ts["Nuevo León"], regression="ct", nlags="auto")
print("KPSS p-value:", p_value)
```

```python
frequencies, power = periodogram(ts["Nuevo León"].to_numpy())
plt.plot(frequencies, power)
plt.xlabel("Frequency")
plt.ylabel("Power")
plt.title("Periodogram - Automobiles NL")
plt.show()
frequencies = frequencies[1:]
power = power[1:]
freq_dom = frequencies[np.argmax(power)]
L = 1 / freq_dom
print("Dominant spectral period (L):", L)
```
